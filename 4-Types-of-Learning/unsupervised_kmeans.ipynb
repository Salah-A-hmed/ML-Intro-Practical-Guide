```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

X = np.array([[20,20000],[25,30000],[40,50000],[22,18000],[35,45000],[50,80000]])

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

plt.scatter(X[:,0], X[:,1], c=kmeans.labels_)
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```
