```python
import numpy as np

grid = np.zeros((3,3))
reward = np.zeros((3,3))
reward[2,2] = 1

q = np.zeros((3,3,4))  # actions: up,down,left,right
alpha = 0.1
gamma = 0.9

for episode in range(50):
    x,y = 0,0
    while (x,y) != (2,2):
        action = np.random.randint(0,4)
        nx, ny = x, y

        if action == 0 and x>0: nx -= 1
        if action == 1 and x<2: nx += 1
        if action == 2 and y>0: ny -= 1
        if action == 3 and y<2: ny += 1

        q[x,y,action] += alpha * (reward[nx,ny] + gamma*np.max(q[nx,ny]) - q[x,y,action])
        x,y = nx,ny

print("Q-table:")
print(q)
```
