```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

df = pd.read_csv("synthetic_points.csv")

X = df[["x","y"]]
y = df["label"]

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)

# visualize data
for label, color in [(0,'blue'),(1,'red')]:
    subset = df[df["label"]==label]
    plt.scatter(subset["x"], subset["y"], c=color)

plt.xlabel("x")
plt.ylabel("y")
plt.title("KNN Points")
plt.show()

print("Predict for point (2,2):", knn.predict([[2,2]])[0])
```
