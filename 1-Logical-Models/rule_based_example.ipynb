```python
import pandas as pd

df = pd.read_csv("weather_play.csv")

def rule_based(row):
    if row["Outlook"] == "Overcast":
        return "Yes"
    if row["Outlook"] == "Sunny" and row["Humidity"] == "High":
        return "No"
    if row["Outlook"] == "Rain" and row["Wind"] == "Weak":
        return "Yes"
    return "No"

df["Predicted"] = df.apply(rule_based, axis=1)

accuracy = (df["Predicted"] == df["Play"]).mean()
print(df)
print("\nAccuracy:", accuracy)
```
